# Identification of Van-der-Pol oscillator system

```python
# import all the necessary libraries
import pysindy as ps
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
# The van der pol oscillator system
def van_der_pol(t, y, mu=0.5):
    return [y[1], 
           -(mu * (1-y[0] ** 2) * y[1] + y[0])]
```

```python
# generate the data for testing
t_span = [0, 100]
y0 = [0.3, 0.1]
sol = solve_ivp(van_der_pol, t_span, y0, dense_output=True)
t = np.linspace(0, 100, 10000)
y = sol.sol(t).T
```

```python
feature_names = ['x', 'xdot']
feature_library = ps.PolynomialLibrary(degree=3)
opt = ps.STLSQ(threshold=0.1)
model=ps.SINDy(feature_library = feature_library,
               feature_names=feature_names, 
              optimizer=opt)
model.fit(y,t=t)
model.print()
```

```python
# Simulate the identified system 
sim = model.simulate(y0,t=t)
```

```python
# plot the solution

# displacement
plt.plot(t,y[:,0],'r',alpha=0.7,linewidth=2.5, label="Original Signal")
plt.plot(t,sim[:,0],'k--',linewidth=2.5,label="Identified Signal")
plt.xlabel("t")
plt.ylabel("x")
plt.legend()

plt.show()

# velocity
plt.plot(t,y[:,1],'r',alpha=0.7,linewidth=2.5, label="Original Signal")
plt.plot(t,sim[:,1],'k--',linewidth=2.5,label="Identified Signal")
plt.xlabel("t")
plt.ylabel("xdot")
plt.legend()
plt.show()

plt.plot(t, np.abs(sim[:,0]-y[:,0]))
plt.show()
```
